# 0. Setup

Run this notebook from inside the unzipped `pinns_vis_unc_data/` folder. Every field is a single-array `.npz` (key = `u` / `v` / `p`).

```python
import numpy as np
from pathlib import Path

HERE = Path.cwd()   # run this notebook from inside pinns_vis_unc_data/
```

# 1. Ground truth

`reference/` holds the clean fields `u, v, p` (each `(n_time, ny, nx) = (200, 50, 100)`) and `grid.npz` with the spatial grid + time values.

```python
u_ref = np.load(HERE / 'reference/u.npz')['u']
v_ref = np.load(HERE / 'reference/v.npz')['v']
p_ref = np.load(HERE / 'reference/p.npz')['p']
grid = np.load(HERE / 'reference/grid.npz')
x_grid, y_grid, t = grid['x_grid'], grid['y_grid'], grid['t']

print('u_ref :', u_ref.shape)
print('v_ref :', v_ref.shape)
print('p_ref :', p_ref.shape)
print('x_grid:', x_grid.shape, ' y_grid:', y_grid.shape, ' t:', t.shape)
```

# 2. Observations

`observations/{regime}/{scenario}/` holds the (corrupted) fields the models were trained on. The `uv` regime has no `p.npz`.

```python
# uvp regime: u, v, p all observed
obs_u = np.load(HERE / 'observations/uvp/bias_cos0.3/u.npz')['u']
obs_v = np.load(HERE / 'observations/uvp/bias_cos0.3/v.npz')['v']
obs_p = np.load(HERE / 'observations/uvp/bias_cos0.3/p.npz')['p']
print('observations/uvp/bias_cos0.3:  u', obs_u.shape, ' v', obs_v.shape, ' p', obs_p.shape)

# uv regime: only u, v observed (no p.npz)
obs_u_uv = np.load(HERE / 'observations/uv/noise0.05/u.npz')['u']
obs_v_uv = np.load(HERE / 'observations/uv/noise0.05/v.npz')['v']
print('observations/uv/noise0.05:     u', obs_u_uv.shape, ' v', obs_v_uv.shape, ' (no p)')
```

# 3. Pred

One run is a folder `predictions/{regime}/{scenario}/lw{w}/` with `u.npz`, `v.npz`, `p.npz`.

```python
run = 'predictions/uvp/bias_cos0.3/lw0.20'
u_pred = np.load(HERE / run / 'u.npz')['u']
v_pred = np.load(HERE / run / 'v.npz')['v']
p_pred = np.load(HERE / run / 'p.npz')['p']
print('run:', run)
print('  u', u_pred.shape, ' v', v_pred.shape, ' p', p_pred.shape)
```
